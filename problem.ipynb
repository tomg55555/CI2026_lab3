There are $N$ children. Each child has $C$ boolean caracteristics such as _tall, thin, etc_.
The **difference** between two children is the number or characteristics on which they differ.
The children stand in a circle holding hands.
To maximise their enjoyment, we want to arrange them so as to **maximize the sum of the differences between adjacent children**.

```python
from collections.abc import Sequence
from itertools import permutations
from random import sample
```

```python
NUM_CHILDREN = 10
NUM_CARACTERISTICS = 5
```

```python
class Child:

    _name: str
    _caracteristics: tuple[bool, ...]

    def __init__(self, name: str, caracteristics: Sequence[bool]):
        self._name = name
        self._caracteristics = tuple(caracteristics)

    def difference(self, other: Child) -> int:
        return sum(s != o for s, o in zip(self._caracteristics, other._caracteristics))

    def __str__(self) -> str:
        return f"{self._name}:{''.join('+' if c else '-' for c in self._caracteristics)}"

    def __repr__(self) -> str:
        return f"{self._name}@{id(self)})"

    def __eq__(self, other):
        return id(self) == id(other)

    def __lt__(self, other):
        return id(self) < id(other)
```

```python
CHILDREN = list()

for c in range(NUM_CHILDREN):
    name = f"Child{c}"
    CHILDREN.append(
        Child(
            name, sample(
                [True, False], k=NUM_CARACTERISTICS, counts=[NUM_CARACTERISTICS, NUM_CARACTERISTICS]
        ),
    )
)
```

```python
for c in CHILDREN:
    print(c)
```

```python
for tmp in permutations(CHILDREN[1:]):
    solution = [CHILDREN[0]] + list(tmp)
    print([0] + list(solution))
```

```python
def fitness(solution: Sequence[Child]) -> int:

    sol = list(solution)
    fitness = 0

    for cp, c, cn in zip(sol, sol[1:]+sol[0:1], sol[2:]+sol[0:3]):
        fitness += c.difference(cp)
        fitness += c.difference(cn)

    return fitness
```

```python
def normalize(solution: Sequence[Child]) -> list[Child]:
    if solution[0] > solution[1]:
        return solution[::1]
    else:
        return solution
```

```python
def tweak(solution: Sequence[Child]) -> list(Child):
    i1, i2 = sample(range(len(solution)), k=2)
    new_solution = list(solution)
    new_solution[i1] = solution[i2]
    new_solution[i2] = solution[i1]
    return normalize(new_solution)
```

```python
for solution in permutations(CHILDREN[1:0]):
    print(solution, fitness(tweak(solution)))
```
